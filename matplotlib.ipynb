```python
import matplotlib.pyplot as plt
```

LINE CHART :- A Line Chart is a type of graph used to represent data points connected by straight lines. It is mainly used to show trends and changes in data over time.
Example: Showing months and sales Chart

IMPORTANT POINTS ---
1. We use multiple plt.plot() statements to draw multiple lines.
2. label gives a name to each line.
3. plt.legend() identifies each line.
4. Different colors and markers make lines easier to distinguish.
5. Multiple lines are useful for comparing sales, temperatures, student marks, and other datasets.

```python
months = ['Jan', 'Feb', 'March', 'Apr']
sales =[100, 160, 130, 50]

plt.plot(months, sales)
plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales chart")
plt.show()
```

MULTIPLE LINES :- A multiple-line chart is a type of graph used to represent and compare two or more sets of data on the same chart using different lines.
It helps us understand the differences, trends, and changes between multiple datasets.

```python
# multiple lines in chart

months =  ['jan', 'feb', 'march', 'april']
actual = [ 100, 150, 200, 250]
target = [120, 140, 160, 180]

plt.plot(months, actual, marker='o', label='actual')
plt.plot(months, target, marker='s', label='target')

plt.title("Actual vs Target")
plt.xlabel("Actual")
plt.ylabel("Target")
plt.legend()
plt.grid()
plt.show()
```

BAR CHART :- A Bar Chart is a type of graph used to represent and compare different categories of data using rectangular bars. The height or length of each bar represents its value.
Example : category and sales

IMPORTANT POINTS ---
1. Bar charts are used to compare categorical data.
2. Bars can be vertical or horizontal.
3. The height or length of a bar represents the value.
4. Bars are generally separated by spaces.
5. Bar charts are useful for comparing students' marks, product sales, and population across cities.

```python
category = ["Electronics", "Clothing", "Furniture", "Beauty", "Sports"]
sales = [1200, 6000, 4500, 3400, 5000]

bars = plt.bar(category, sales,
               color = "pink",
               width = 0.6,
               edgecolor = "black",
               linewidth = 1)

plt.bar_label(bars)      # ye labels show karti hai 
plt.title("category and Sales Chart")
plt.xlabel("category")
plt.ylabel("sales")
plt.show()
```

AND this is for Horizontal Bar Chart (barh)

```python
student = ["palak", "divya", "priya", "neha"]
marks = [80, 30, 15, 60]

plt.barh(student, marks)

plt.xlabel("Marks")
plt.ylabel("Students")
plt.title("Student's marks")

plt.show()
```

HISTOGRAM CHART :- A Histogram is a type of graph used to represent the frequency distribution of numerical data using adjacent rectangular bars. It shows how data values are distributed across different intervals or bins.
Example: Distribution of students' marks in different ranges.

IMPORTANT POINTS ---
1. A histogram represents the frequency distribution of numerical data.
2. It is mainly used for continuous numerical data.
3. The X-axis represents data intervals, called bins.
4. The Y-axis usually represents frequency.
5. The bars generally touch each other because the intervals are continuous.
6. It helps identify the shape and spread of a data distribution.
7. It can help reveal skewness, clusters, and possible outliers.

```python
marks = [23, 55, 66, 78, 90, 3, 56, 70, 75, 45, 20, 50]

plt.hist(marks, bins= 6, edgecolor='black')

plt.title("distribution of marks ")
plt.xlabel("marks")
plt.ylabel("no. of students")

plt.show()
```

SCATTER CHART :- A Scatter Chart (Scatter Plot) is a type of graph used to represent the relationship between two numerical variables using individual dots or points.
It helps us identify patterns, trends, clusters, and relationships between variables.
Example: Showing the relationship between students' study hours and exam marks.

IMPORTANT POINTS ---
1. A scatter chart displays data using individual dots.
2. It is used to study the relationship between two numerical variables.
3. The X-axis represents one variable, and the Y-axis represents another.
4. It helps identify positive, negative, or no correlation.
5. It can show clusters and possible outliers.
6. It is useful for data analysis and machine learning.
7. Unlike a line chart, a scatter chart does not automatically connect points with lines.


```python
hours = [1, 2, 3, 4, 5, 6]
marks = [30, 40, 55, 65, 77, 87]

plt.scatter(hours, marks, color ='pink')

plt.title("relationship")
plt.xlabel("hours")
plt.ylabel("hours")
plt.grid(alpha = 0.5)
plt.show()
```
