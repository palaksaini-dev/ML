PART 3

```python
# 1. Standardize missing-value markers such as N/A, unknown, blank strings and - wherever appropriate.
missing_values = ["N/A", "n/a", "unknown", "Unknown", "-", ""]

df = df.replace(missing_values, 'nan')
df
```

```python
# 2. Remove unnecessary leading and trailing spaces from text columns.
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
df
```

```python
# 3. Standardize Gender values into consistent categories.
df["Gender"].str.strip().str.lower()

df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})
df
```

```python
# 4. Standardize Subject names, such as Mathematics and math.
df["Subject"].str.strip().str.lower()

df["Subject"].replace({
    "math": "Mathematics",
    "maths": "Mathematics",
    "mathematics": "Mathematics",
    "english": "English",
    "science": "Science",
    "computer": "Computer Science",
    "computer science": "Computer Science"
})
df
```

```python
# 5. Standardize City names with consistent capitalization.
df["City"] = df["City"].str.strip().str.title()
df
```

```python
# 6. Convert Attendance_%, Study_Hours_Per_Day and Exam_Score into numeric data types.
cols = ["Attendance_%", "Study_Hours_Per_Day", "Exam_Score"]

for col in cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[cols].dtypes)
```

```python
# 7. Identify attendance values outside 0–100, negative study hours and exam scores outside 0–100.
print("Invalid attendance:")
print(df[~df["Attendance_%"].between(0, 100)])

print("Negative study hours:")
print(df[df["Study_Hours_Per_Day"] < 0])

print("Invalid exam scores:")
print(df[~df["Exam_Score"].between(0, 100)])
```

```python
# 2nd method
print(df[df["Attendance_%"].notna() & df["Attendance_%"].between(0, 100)])

print(df[df["Study_Hours_Per_Day"].notna() & (df["Study_Hours_Per_Day"] < 0)])

print(df[df["Exam_Score"].notna() & df["Exam_Score"].between(0, 100)])
```

```python
# 8. Handle invalid numerical values using a clearly stated rule.
df.loc[~df["Attendance_%"].between(0, 100), "Attendance_%"]

df.loc[df["Study_Hours_Per_Day"] < 0, "Study_Hours_Per_Day"]

df.loc[~df["Exam_Score"].between(0, 100), "Exam_Score"]
```

```python
# 9. Remove exact duplicate rows, but do not automatically remove every repeated Student_ID.
print("Before:", df.shape)

df = df.drop_duplicates()

print("After:", df.shape)
```

```python
# 10. Convert Exam_Date into a proper datetime data type and identify invalid dates.
df["Exam_Date"] = pd.to_datetime(df["Exam_Date"], errors="coerce")

print(df[df["Exam_Date"].isna()])
```

```python
# 11. Handle missing numerical and categorical values using suitable strategies.
# Numerical input features ko median se fill karo
numeric_features = ["Attendance_%", "Study_Hours_Per_Day"]

for col in numeric_features:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns mein missing values ko Unknown banao
categorical_cols = [
    "Gender", "Class", "Subject", "City", "Parent_Contacted"
]

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")

# Exam_Score aur Exam_Date ko filhaal missing rehne do
print(df.isnull().sum())
```

```python
# 9. Create a new column named Performance_Level using a rule you define and document.
conditions = [
    df["Exam_Score"] >= 80,
    df["Exam_Score"] >= 70,
    df["Exam_Score"] >= 50,
    df["Exam_Score"] < 50
]

labels = [
    "Excellent",
    "Good",
    "Average",
    "Needs Improvement"
]

df["Performance_Level"] = np.select(
    conditions,
    labels,
    default="Unknown"
)

print(df[["Exam_Score", "Performance_Level"]].head(10))
```
